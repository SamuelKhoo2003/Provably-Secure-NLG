```python
from certifiable_learning_stability.dpa_certifier import HalfmoonsCertifier, MnistCertifier, Cifar10Certifier, BlobsCertifier
from certifiable_learning_stability.rdp_certifier import Cifar10Certifier as Cifar10RDPCertifier
from certifiable_learning_stability.certification_methods import CertificationMethod, AggregationType
from experiments.reproducibility import make_reproducible, get_device
from experiments.misc import dummy_hparams_agt, dummy_hparams_hrdp, dummy_hparams_prdp, dummy_hparams_sgd
import torch
```

```python
SEED = 42
make_reproducible(SEED)
```

# HALFMOONS

```python

hp_agt_halfmoons = {
    "epochs": 4,
    "batch_size": 5000,
    "lr": 1,
    "lr_decay": 0.6,
    "lr_min": 1e-3,
    "weight_decay": 1e-4,
    "ks_private": list(range(1, 33, 4)),
    "clip_gammas": [0.1] # Seems to be optimal
}
hp_hrdp_halfmoons = dummy_hparams_hrdp()
hp_prdp_halfmoons = dummy_hparams_prdp()
hp_sgd_halfmoons = dummy_hparams_sgd()
hp_bag_halfmoons = {
    "num_partitions": 250,
    "test_batch_size": 500,
    "seed": SEED,
    "method_name": "bag_agt_halfmoons",
    "hp_sgd": hp_sgd_halfmoons,
    "hp_agt": hp_agt_halfmoons,
    "hp_hrdp": hp_hrdp_halfmoons,
    "hp_prdp": hp_prdp_halfmoons
}
device = get_device(index=1)
```

```python
kwargs = {"logfile_name": "generalized_framework", "write_to_file": True, "save": True}
halfmoons_dpa_certifier = HalfmoonsCertifier(hp_bag_halfmoons, device, save_kwargs=kwargs)
print(halfmoons_dpa_certifier.save_load_dir)
```

```python
halfmoons_dpa_certifier.train_dpa_partitions((CertificationMethod.AGT,), partitioning_method="bag")
```

```python
# 1000 randomly sampled points from the test set
test_set_subset = torch.utils.data.Subset(halfmoons_dpa_certifier.test_set, indices=torch.randperm(len(halfmoons_dpa_certifier.test_set))[:1000])

halfmoons_dpa_certifier.agt_bagging_guarantee(9, test_set=test_set_subset)
```

# BLOBS

```python
hp_agt_blobs = {
    "epochs": 4,
    "batch_size": 400,
    "lr": 2.0,
    "lr_decay": 0.6,
    "lr_min": 1e-3,
    "ks_private": list(range(0, 101, 5)),
    "clip_gammas": [0.06] # Seems to be optimal
}
hp_hrdp_blobs = dummy_hparams_hrdp()
hp_prdp_blobs = dummy_hparams_prdp()
hp_sgd_blobs = dummy_hparams_sgd()
hp_bag_blobs = {
    "num_partitions": 250,
    "test_batch_size": 500,
    "seed": SEED,
    "method_name": "bag_agt_blobs",
    "hp_sgd": hp_sgd_blobs,
    "hp_agt": hp_agt_blobs,
    "hp_hrdp": hp_hrdp_blobs,
    "hp_prdp": hp_prdp_blobs
}
device = get_device(index=1)
```

```python
kwargs = {"logfile_name": "generalized_framework", "write_to_file": True, "save": True}
blobs_dpa_certifier = BlobsCertifier(hp_bag_blobs, device, save_kwargs=kwargs)
print(blobs_dpa_certifier.save_load_dir)
```

```python
blobs_dpa_certifier.train_dpa_partitions((CertificationMethod.AGT,), partitioning_method="bag")
```

```python
# 1000 randomly sampled points from the test set
test_set_subset = torch.utils.data.Subset(blobs_dpa_certifier.test_set, indices=torch.randperm(len(blobs_dpa_certifier.test_set))[:1000])

for k_private in hp_agt_blobs["ks_private"]:
    blobs_dpa_certifier.agt_bagging_guarantee(k_private, clip_gamma=hp_agt_blobs["clip_gammas"][0], test_set=test_set_subset)
```
