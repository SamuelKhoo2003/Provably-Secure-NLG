```python
from certifiable_learning_stability.rdp_certifier import HalfmoonsCertifier, MnistCertifier, Cifar10Certifier
from certifiable_learning_stability.threats import Threats
from datasets.cifar import subset
from experiments.reproducibility import make_reproducible, get_device
```

```python
SEED = 42
make_reproducible(SEED)
device = get_device(index=0)
```

## Halfmoons

```python
hyperparams = {
    "epochs": 2,
    "batch_size": 160,
    "lr": 0.02,
    "mechanism_samples": 1000,
    "confidence": 0.99,
    "seed": SEED,
    "sigma": 1.5,
    "alpha": 2.1,
    "max_grad_norm": 0.5,
    "weight_decay": 0.001,
}
kwargs = {"logfile_name": "tests", "write_to_file": True, "save": False}
halfmoons_rdp_certifier = HalfmoonsCertifier(hyperparams, device, save_kwargs=kwargs)
```

```python
halfmoons_rdp_certifier.certify_params(hit_ratios=[0.7, 0.75, 0.8, 0.9, 0.95, 0.98, 0.99])
```

```python
halfmoons_rdp_certifier.certify_points()
```
